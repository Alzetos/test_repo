```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
movie_table = np.array([[40, 10],
                       [15, 35]])
chi2, p_value, dof, expected = stats.chi2_contingency(movie_table)
print(chi2) # статистика хи-квадрат
print(p_value)
print(dof) #число степеней свободы
print(expected) #ожидаемая матрица
```

```python
battery_life = [12, 15, 14, 11, 24, 13, 14, 12, 28, 14]
statistics, p_value = stats.shapiro(battery_life)
if (p_value >= 0.05):
    print("Нормальное распредление")
else: 
    print("Нет нормального распределения")
print(f"p-value = {p_value}\n")
print(f"statistics = {statistics}")
```

```python
battery_life = [12, 15, 14, 11, 24, 13, 14, 12, 28, 14]

stats.probplot(battery_life, dist="norm", plot=plt)
plt.title("Q-Q PLOT для времени работы аккумулятора")
plt.xlabel("Теоретические квантили")
plt.ylabel("Реальные квантили")
plt.grid(True)
plt.show()
```

```python
data = {
    'часы_работы': [1, 2, 3, 4, 5, 6, 7, 8],
    'выпущено_деталей': [10, 20, 30, 40, 50, 60, 70, 0]
}
df = pd.DataFrame(data)
pearson_matrix = df.corr()
print(pearson_matrix)
```

```python
sns.heatmap(pearson_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.show
```

```python
data = {
    'часы_работы': [1, 2, 3, 4, 5, 6, 7, 8],
    'выпущено_деталей': [10, 20, 30, 40, 50, 60, 70, 0]
}
df = pd.DataFrame(data)
print(df.describe())

median_detail = df["выпущено_деталей"].median()
df["выпущено_деталей"] = df["выпущено_деталей"].replace(0, median_detail)

df["ранг_часов"] = df["часы_работы"].rank()
df["ранг_деталей"] = df["выпущено_деталей"].rank()
mask = ["ранг_деталей","ранг_часов"]

spirson_matrix = df[mask].corr()
print(spirson_matrix)

sns.heatmap(spirson_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.show()
```

```python
data = np.random.normal(loc=0, scale=1, size=1000)

statistics, p_value, = stats.ks_1samp(data, stats.norm.cdf)
print(f"D: {statistics}")
print(f"p-value: {p_value}") 
```
