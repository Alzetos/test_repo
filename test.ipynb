```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
movie_table = np.array([[40, 10],
                       [15, 35]])
chi2, p_value, dof, expected = stats.chi2_contingency(movie_table)
print(chi2) # статистика хи-квадрат
print(p_value)
print(dof) #число степеней свободы
print(expected) #ожидаемая матрица
```

```python
battery_life = [12, 15, 14, 11, 24, 13, 14, 12, 28, 14]
statistics, p_value = stats.shapiro(battery_life)
if (p_value >= 0.05):
    print("Нормальное распредление")
else: 
    print("Нет нормального распределения")
print(f"p-value = {p_value}\n")
print(f"statistics = {statistics}")
```

```python
battery_life = [12, 15, 14, 11, 24, 13, 14, 12, 28, 14]

stats.probplot(battery_life, dist="norm", plot=plt)
plt.title("Q-Q PLOT для времени работы аккумулятора")
plt.xlabel("Теоретические квантили")
plt.ylabel("Реальные квантили")
plt.grid(True)
plt.show()
```

```python
data = {
    'часы_работы': [1, 2, 3, 4, 5, 6, 7, 8],
    'выпущено_деталей': [10, 20, 30, 40, 50, 60, 70, 0]
}
df = pd.DataFrame(data)
pearson_matrix = df.corr()
print(pearson_matrix)
```
